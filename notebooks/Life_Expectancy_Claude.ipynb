Because this data is a times series which can be regarded as a sequence of values, imputation can be used to estimate new values for the missing values some of these features contain, specifically the big null value offenders.  

```python
global df
# Because this is a time series (each country tracked year over year),
# interpolate within each country first -- this fills interior gaps
# using that country's own neighboring years, which is more meaningful
# than a blanket column-wide fill. Any values interpolation can't reach
# (e.g. a country missing data at the very start/end of its series) fall
# back to that country's own median, then to the overall column median
# as a last resort for a country with no data at all for a feature.
numeric_cols = [c for c in feature_vars if c != 'Life_Expectancy']
df = df.sort_values(['Country', 'Year'])

for col in numeric_cols:
    df[col] = df.groupby('Country')[col].transform(lambda s: s.interpolate(limit_direction='both'))
for col in numeric_cols:
    df[col] = df[col].fillna(df.groupby('Country')[col].transform('median'))
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

na_values = df.isna().sum().reset_index()
na_values.columns = ["Features", "Missing_Values"]
na_values["Missing_Percent"] = round(na_values.Missing_Values / len(df) * 100, 2)
print("Summarize NULLS after imputation: \n", na_values[na_values.Missing_Values > 0])
print("\nTotal null cells remaining:", df[numeric_cols].isnull().sum().sum())
```

I would like to know the outlier details by feature so I will start there and apply Tukey’s rule which is a statistical method for identify what constitutes an outlier. 

```python
#outlier_summary()  # Count outliers and plot
with open('charts/Outlier_Summary.txt', mode='w', newline='') as outlier_file:
    outlier_writer = csv.writer(outlier_file, delimiter=',', quotechar='"', quoting=csv.QUOTE_MINIMAL)
    for col in feature_vars:
        q75, q25 = np.percentile(df[col], [75, 25])
        iqr = q75 - q25
        min_val = q25 - (iqr * 1.5)
        max_val = q75 + (iqr * 1.5)
        count = len(np.where((df[col] > max_val) | (df[col] < min_val))[0])
        percent = round(count / len(df[col]) * 100, 2)

        total_chars = 45
        len_first_chars = 7
        len_col = len(col)
        chars_needed = total_chars - len_col - len_first_chars

        print(len_first_chars * '-' + col + ' Outliers' + chars_needed * '-')
        print('Count: {}'.format(count))
        print('Percentage of Data: {}%'.format(percent))

        outlier_writer.writerow([len_first_chars * '-' + col + ' Outliers' + chars_needed * '-'])
        outlier_writer.writerow(['Count: {}'.format(count)])
        outlier_writer.writerow(['Percentage of Data: {}%'.format(percent)])
        outlier_writer.writerow(' ')
```

### WINSORIZATION
Let's work with the outliers.

Because there are a number of features with outlier issues and each one has its own unique issues, the method I will use to address outliners is to Winsorize the data.  This method will help to minimize that influence that outliers have on each variable.  Winsorizing the data will enable me to set upper and lower limits on each feature.  This limits the outlier’s severity and influence on the data.  

The parameters for winsorizing will be stored in a dictionary for easy access later in the project.


```python
# Set windsorization vars with limits
def check_wins_plot(col, lower_limit=0, upper_limit=0, show_plot=True):
    global wins_dict
    wins_data = winsorize(df[col], limits=(lower_limit, upper_limit))  # Create winsorized col entries
    wins_dict[col] = wins_data  # Add the winsorized data to the wins dictionary

    if show_plot == True:
        plt.figure(figsize=(10, 5))
        plt.subplot(1, 2, 1)
        plt.boxplot(df[col])
        plt.title('Original Data {}'.format(col), fontweight='bold')
        plt.subplot(1, 2, 2)
        plt.boxplot(wins_data)
        pyplot.suptitle('Compare Before/After Winsorized\n', fontsize=16, verticalalignment='top',
                        horizontalalignment='center', fontweight='bold')
        plt.title('Winsorised Data=({},{}) {}'.format(lower_limit, upper_limit, col), fontweight='bold')
        plt.savefig('charts/Compare_Winsorized' + col + '.png', dpi=None, facecolor='w', edgecolor='g',
                    orientation='landscape',
                    format=None, transparent=False, bbox_inches=None, pad_inches=0.10, metadata=None)
        plt.show()

# FIX: named-column dict instead of positional feature_vars[i] indices.
WINSORIZE_LIMITS = {
    'Life_Expectancy':          dict(lower_limit=.01, show_plot=True),
    'Adult_Mortality':          dict(upper_limit=.04),
    'Infant_Deaths':            dict(upper_limit=.05),
    'Alcohol':                  dict(upper_limit=.0025),
    'Pct_Expenditure':          dict(upper_limit=.135),
    'Hepatitis_B':              dict(lower_limit=.1),
    'Measles':                  dict(upper_limit=.19),
    'Under-Five_Deaths':        dict(upper_limit=.05),
    'Polio':                    dict(lower_limit=.1),
    'Ttl_Expend':               dict(upper_limit=.02),
    'Diphtheria':               dict(lower_limit=.105),
    'HIV_AIDS':                 dict(upper_limit=.185, show_plot=True),
    'GDP':                      dict(upper_limit=.105),
    'Population':               dict(upper_limit=.07),
    'Thinness_10-19_Years':     dict(upper_limit=.035),
    'Thinness_5-9_Years':       dict(upper_limit=.035),
    'Income_Comp_Of_Resources': dict(lower_limit=.05),
    'Schooling':                dict(lower_limit=.025, upper_limit=.005),
}

for col in feature_vars:
    check_wins_plot(col, **WINSORIZE_LIMITS[col])
```

The boxplots above are illustrations of successful winsorization for life expectancy and HIV/AIDS. These two are just shown as examples of the winsorization results for all numerical features.

Polynomial regression can be beneficial in some cases where the data may have a curvilinear relationship between the target variable and the independent variables. Below, several polynomial degrees are compared on a held-out test set (rather than only checking training fit), since higher-degree polynomials can look better and better on training data while getting worse at predicting new data -- overfitting.

```python
# Polynomial Regression: Schooling -> Life Expectancy
poly_df = wins_df.dropna().copy()  # Copy winsorized dataframe
x = poly_df.Schooling.values.reshape(-1, 1)
y = poly_df['Life_Expectancy'].values

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.20, random_state=42)  # Split data

print("Comparing polynomial degrees on a HELD-OUT test set (not just training fit):\n")
for degree in [1, 2, 3, 5, 8, 15]:
    poly = PolynomialFeatures(degree=degree)
    x_train_poly = poly.fit_transform(x_train)
    x_test_poly = poly.transform(x_test)
    regr = LinearRegression()
    regr.fit(x_train_poly, y_train)
    train_r2 = r2_score(y_train, regr.predict(x_train_poly))
    test_r2 = r2_score(y_test, regr.predict(x_test_poly))
    flag = "  <- overfitting: test R2 drops well below train R2" if train_r2 - test_r2 > 0.05 else ""
    print(f"  degree={degree:2d}   train R2={train_r2:.4f}   test R2={test_r2:.4f}{flag}")

# Degree 3 gives nearly all of the benefit without overfitting -- plot that one.
best_degree = 3
poly_features = PolynomialFeatures(degree=best_degree)
level_poly = poly_features.fit_transform(x_train)
regr = LinearRegression()
regr.fit(level_poly, y_train)

x_line = np.linspace(min(x), max(x), 200).reshape(-1, 1)
y_line = regr.predict(poly_features.transform(x_line))

plt.scatter(x_train, y_train, color="blue", alpha=0.4, label="Train")
plt.scatter(x_test, y_test, color="orange", alpha=0.6, label="Test")
plt.plot(x_line, y_line, color="red", label=f"Degree {best_degree} fit")
plt.xlabel("Schooling")
plt.ylabel("Life Expectancy")
plt.title("Polynomial Regression - Schooling & Life Expectancy", fontweight='bold', fontsize=13)
plt.legend()
plt.savefig('charts/Polynomial_Regression.png', dpi=None, facecolor='w', edgecolor='g', orientation='landscape',
            format=None, transparent=False, bbox_inches=None, pad_inches=0.10, metadata=None)
plt.show()
```

Comparing several polynomial degrees on a held-out test set (rather than only training fit) shows results leveling off around degree 3. Higher degrees -- like the original degree 15 -- add complexity without meaningfully improving, and eventually hurting, test-set performance. That's the classic signature of overfitting.

### Logistic Regression

```python
# Logistic Regression: classify High vs Low Life Expectancy
logi_df = wins_df.dropna().copy()  # Copy winsorized dataframe
predictor_cols = [c for c in feature_vars if c != 'Life_Expectancy']

median_le = logi_df['Life_Expectancy'].median()
logi_df['LE_Class'] = (logi_df['Life_Expectancy'] >= median_le).astype(int)  # 1 = High, 0 = Low
print(f"Splitting at the median Life Expectancy of {median_le:.1f} years")
print(logi_df['LE_Class'].value_counts())

X_data = logi_df[predictor_cols]
X = (X_data - X_data.min()) / (X_data.max() - X_data.min())  # Normalize
y = logi_df['LE_Class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

logi = LogisticRegression(solver="liblinear")
logi_model = logi.fit(X_train, y_train)  # The ideal fit line

print("\nlogi_model.intercept_ :", logi_model.intercept_)
print("logi_model.coef_ (higher magnitude = stronger influence on High/Low classification):")
for name, coef in zip(predictor_cols, logi_model.coef_[0]):
    print(f"  {name}: {coef:.4f}")
```

T-TEST

```python
# Ttest to determine significance of difference between developed and developing
ttest_results = stats.ttest_ind(wins_df.loc[wins_df['Status'] == 'Developed', 'Life_Expectancy'],
                                 wins_df.loc[wins_df['Status'] == 'Developing', 'Life_Expectancy'])
print("t-statistic: {:.3f}".format(ttest_results.statistic))
print("p-value: {:.3e}".format(ttest_results.pvalue))
```

The t-test results show that there is a significant difference between developing and developed countries when it comes to life expectancy.
